```python
import pandas as pd
from pathlib import Path

# Dataset folder path
data_path = Path("../data/raw")

# Check available files
print("Files available:")
for file in data_path.iterdir():
    print(file.name)
```

```python
# Load all three datasets
netflix_df = pd.read_csv(data_path / "netflix_titles.csv")
movies_df = pd.read_csv(data_path / "movies.csv")
shows_df = pd.read_csv(data_path / "shows.csv")

print("All datasets loaded successfully!")
```

```python
# Check dataset dimensions
print("NETFLIX TITLES")
print("Rows and Columns:", netflix_df.shape)
print("Columns:", netflix_df.columns.tolist())

print("\nMOVIES")
print("Rows and Columns:", movies_df.shape)
print("Columns:", movies_df.columns.tolist())

print("\nTV SHOWS")
print("Rows and Columns:", shows_df.shape)
print("Columns:", shows_df.columns.tolist())
```

```python
# Check missing values
print("NETFLIX TITLES - Missing Values")
print(netflix_df.isnull().sum())

print("\nMOVIES - Missing Values")
print(movies_df.isnull().sum())

print("\nTV SHOWS - Missing Values")
print(shows_df.isnull().sum())

# Check duplicate rows
print("\nDuplicate Records:")
print("Netflix Titles:", netflix_df.duplicated().sum())
print("Movies:", movies_df.duplicated().sum())
print("TV Shows:", shows_df.duplicated().sum())
```

```python
# Inspect duplicate rows in Movies
print("MOVIES DUPLICATES")
display(movies_df[movies_df.duplicated(keep=False)])

# Inspect duplicate rows in TV Shows
print("TV SHOWS DUPLICATES")
display(shows_df[shows_df.duplicated(keep=False)])
```

```python
# Remove rows where title is missing
movies_df = movies_df.dropna(subset=["title"]).reset_index(drop=True)
shows_df = shows_df.dropna(subset=["title"]).reset_index(drop=True)

# Check updated dataset sizes
print("Movies shape:", movies_df.shape)
print("TV Shows shape:", shows_df.shape)

# Check duplicates again
print("Movies duplicates:", movies_df.duplicated().sum())
print("TV Shows duplicates:", shows_df.duplicated().sum())
```

```python
# Calculate missing value percentage
print("MOVIES - Missing Value Percentage")
display((movies_df.isnull().mean() * 100).round(2))

print("TV SHOWS - Missing Value Percentage")
display((shows_df.isnull().mean() * 100).round(2))
```

```python
print("MOVIES DATA TYPES")
print(movies_df.dtypes)

print("\nTV SHOWS DATA TYPES")
print(shows_df.dtypes)

print("\nAVAILABLE GLOBALLY VALUES")
print("Movies:", movies_df["available_globally"].unique())
print("TV Shows:", shows_df["available_globally"].unique())
```

```python
# Remove repeated header rows
movies_df = movies_df[
    movies_df["available_globally"] != "Available Globally?"
].reset_index(drop=True)

shows_df = shows_df[
    shows_df["available_globally"] != "Available Globally?"
].reset_index(drop=True)

# Verify
print("Movies:", movies_df.shape)
print("TV Shows:", shows_df.shape)

print("\nMovies values:")
print(movies_df["available_globally"].value_counts(dropna=False))

print("\nTV Shows values:")
print(shows_df["available_globally"].value_counts(dropna=False))
```

```python
# Convert release_date into datetime
movies_df["release_date"] = pd.to_datetime(
    movies_df["release_date"], errors="coerce"
)

shows_df["release_date"] = pd.to_datetime(
    shows_df["release_date"], errors="coerce"
)

# Verify data types
print("Movies release_date:", movies_df["release_date"].dtype)
print("TV Shows release_date:", shows_df["release_date"].dtype)

# Preview dates
display(movies_df[["title", "release_date"]].head())
display(shows_df[["title", "release_date"]].head())
```

```python
print("MOVIES RUNTIME VALUES")
print(movies_df["runtime"].dropna().unique()[:20])

print("\nTV SHOWS RUNTIME VALUES")
print(shows_df["runtime"].dropna().unique()[:20])
```

```python
import re

def convert_runtime(value):
    if pd.isna(value):
        return None

    hours = re.search(r'(\d+)H', value)
    minutes = re.search(r'(\d+)M', value)

    h = int(hours.group(1)) if hours else 0
    m = int(minutes.group(1)) if minutes else 0

    return h * 60 + m


movies_df["runtime_minutes"] = movies_df["runtime"].apply(convert_runtime)
shows_df["runtime_minutes"] = shows_df["runtime"].apply(convert_runtime)

print("MOVIES")
display(movies_df[["title", "runtime", "runtime_minutes"]].head())

print("TV SHOWS")
display(shows_df[["title", "runtime", "runtime_minutes"]].head())
```

```python
# Check engagement metrics
print("MOVIES")
print(movies_df[["hours_viewed", "views"]].describe())

print("\nTV SHOWS")
print(shows_df[["hours_viewed", "views"]].describe())

# Check missing values
print("\nMissing engagement values")
print("Movies:")
print(movies_df[["hours_viewed", "views"]].isnull().sum())

print("TV Shows:")
print(shows_df[["hours_viewed", "views"]].isnull().sum())
```

```python
# Remove rows with missing engagement metrics
movies_df = movies_df.dropna(
    subset=["hours_viewed", "views"]
).reset_index(drop=True)

shows_df = shows_df.dropna(
    subset=["hours_viewed", "views"]
).reset_index(drop=True)

# Verify
print("Movies shape:", movies_df.shape)
print("TV Shows shape:", shows_df.shape)

print("\nMissing engagement values:")
print("Movies:", movies_df[["hours_viewed", "views"]].isnull().sum().sum())
print("TV Shows:", shows_df[["hours_viewed", "views"]].isnull().sum().sum())
```

```python
# Check missing values in Netflix catalog
print("Missing Values:")
print(netflix_df.isnull().sum())

print("\nMissing Percentage:")
print((netflix_df.isnull().mean() * 100).round(2))

print("\nContent Type Distribution:")
print(netflix_df["type"].value_counts())

print("\nRating Distribution:")
print(netflix_df["rating"].value_counts().head(10))
```

```python
# Fill missing categorical information
columns_to_fill = ["director", "cast", "country", "rating", "duration"]

for col in columns_to_fill:
    netflix_df[col] = netflix_df[col].fillna("Unknown")

# Convert date_added into datetime
netflix_df["date_added"] = pd.to_datetime(
    netflix_df["date_added"],
    errors="coerce"
)

# Verify missing values
print("Remaining Missing Values:")
print(netflix_df.isnull().sum())

print("\nUpdated Data Types:")
print(netflix_df.dtypes)
```

```python
# Reload original date values
original_dates = pd.read_csv(
    data_path / "netflix_titles.csv"
)["date_added"]

# Clean spaces and convert dates
clean_dates = original_dates.astype("string").str.strip()

netflix_df["date_added"] = pd.to_datetime(
    clean_dates,
    format="%B %d, %Y",
    errors="coerce"
)

# Verify
print("Missing dates:", netflix_df["date_added"].isna().sum())
print("\nSample dates:")
display(netflix_df[["title", "date_added"]].head(10))
```

```python
print("Netflix Dataset Shape:", netflix_df.shape)

print("\nMissing Values:")
display(netflix_df.isnull().sum())

print("\nData Types:")
display(netflix_df.dtypes)
```

```python
print("MOVIES REPORT PERIODS")
display(movies_df["report"].value_counts())

print("\nTV SHOWS REPORT PERIODS")
display(shows_df["report"].value_counts())
```

```python
print("MOVIES")
print("Total Rows:", len(movies_df))
print("Unique Titles:", movies_df["title"].nunique())
print("Repeated Title Rows:", movies_df["title"].duplicated().sum())

print("\nTV SHOWS")
print("Total Rows:", len(shows_df))
print("Unique Titles:", shows_df["title"].nunique())
print("Repeated Title Rows:", shows_df["title"].duplicated().sum())
```

```python
print("MOVIES SUMMARY")
display(
    movies_df[
        ["hours_viewed", "views", "runtime_minutes"]
    ].describe().round(2)
)

print("\nTV SHOWS SUMMARY")
display(
    shows_df[
        ["hours_viewed", "views", "runtime_minutes"]
    ].describe().round(2)
)
```

```python
display(
    movies_df[
        movies_df["runtime_minutes"] > 300
    ][
        ["title", "report", "runtime", "runtime_minutes",
         "hours_viewed", "views"]
    ].sort_values(
        "runtime_minutes", ascending=False
    ).head(15)
)
```

```python
movies_df["runtime_flag"] = movies_df["runtime_minutes"].apply(
    lambda x: "Review" if x > 300 else "Normal"
)

shows_df["runtime_flag"] = shows_df["runtime_minutes"].apply(
    lambda x: "Review" if x > 3000 else "Normal"
)

print("MOVIES")
display(movies_df["runtime_flag"].value_counts())

print("\nTV SHOWS")
display(shows_df["runtime_flag"].value_counts())
```

```python
movies_df["calculated_hours"] = (
    movies_df["views"] * movies_df["runtime_minutes"] / 60
)

shows_df["calculated_hours"] = (
    shows_df["views"] * shows_df["runtime_minutes"] / 60
)

print("MOVIES")
display(
    movies_df[
        ["title", "hours_viewed", "calculated_hours"]
    ].head(10).round(2)
)

print("\nTV SHOWS")
display(
    shows_df[
        ["title", "hours_viewed", "calculated_hours"]
    ].head(10).round(2)
)
```

```python
datasets = {
    "Netflix Catalog": netflix_df,
    "Movies Engagement": movies_df,
    "TV Shows Engagement": shows_df
}

for name, df in datasets.items():
    print(f"\n{name}")
    print("Rows:", df.shape[0])
    print("Columns:", df.shape[1])
    print("Missing Values:", df.isnull().sum().sum())
```

```python
movies_df.drop(
    columns=["calculated_hours", "runtime_flag"],
    inplace=True
)

shows_df.drop(
    columns=["calculated_hours", "runtime_flag"],
    inplace=True
)

print("Movies:", movies_df.shape)
print("TV Shows:", shows_df.shape)
```

```python
from pathlib import Path

processed_path = Path("../data/processed")
processed_path.mkdir(parents=True, exist_ok=True)

netflix_df.to_csv(processed_path / "netflix_titles_cleaned.csv", index=False)
movies_df.to_csv(processed_path / "movies_cleaned.csv", index=False)
shows_df.to_csv(processed_path / "shows_cleaned.csv", index=False)

print("All 3 cleaned datasets saved successfully!")
```
